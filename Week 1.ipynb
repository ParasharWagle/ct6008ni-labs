```sql
SELECT order_status, COUNT(*) AS num_orders
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
GROUP BY order_status
```

```sql
SELECT month(order_purchase_timestamp) AS order_month,
       COUNT(*) AS num_orders
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
GROUP BY month(order_purchase_timestamp)
ORDER BY order_month
```

```sql
SELECT avg(datediff(order_delivered_customer_date, order_purchase_timestamp)) AS avg_days
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
WHERE order_status = 'delivered'
```

```sql
SELECT date(order_purchase_timestamp) AS order_day, COUNT(*) AS num_orders
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
GROUP BY date(order_purchase_timestamp)
ORDER BY num_orders DESC
LIMIT 1
```

```sql
SELECT COUNT(*) AS late_orders
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
WHERE order_status = 'delivered'
  AND order_delivered_customer_date > order_estimated_delivery_date
```

```sql
SELECT month(order_purchase_timestamp) AS order_month,
       avg(CASE WHEN order_status = 'canceled' THEN 1 ELSE 0 END) AS cancel_rate
FROM workspace.default.orders_2025_e_17_ce_22_e_20_cb_4777_872_b_0_e_7_e_0071_b_31_c_213124
GROUP BY month(order_purchase_timestamp)
ORDER BY cancel_rate DESC
```
